```python
from sklearn.tree import DecisionTreeClassifier

#Input
X=[[1],[2],[5],[6]] #Hours of study as Input
y=['fail','fail','pass','pass'] #Output or Result

#Create an Object of the DesicsionTreeClassifier statistical algorithm.
model = DecisionTreeClassifier()
#Training the Model.
model.fit(X,y)

#Now Predicting new Input
result =model.predict([[input("Enter Study Hours:")]])
#Displaying the result here
print('Predicted Result :',result[0])
```

create model

```python

from sklearn.tree import DecisionTreeClassifier
import pandas as pd
#Here we are dividing the entire Data into training or testing purpose.
from sklearn.model_selection import train_test_split
#We can save the model locally for later use.
import joblib #Comes with scikit-learn
#Reading from the csv file
df = pd.read_csv("./documents/students_result.csv")
#Displaying all Data from csv file
#print(df)

#Split into Input and Output.
X= df[["hours_studied","attendance"]] #Input
print(X)

y= df['result']
print(y)

#Never fit all Data to Model , rather create 2 catagories
#Training Data --> 70% 
#Testing Data  --> 30% unseen data

X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42 #It will keep same randomness for each time we run the model.
)
#X_train --> 70% Training Data
#X_test  --> 30% Testing Data
#y_train --> 70% result Data
#y_test --> 30% result data 

print("X_train")
print(len(X_train))
print(X_train)

#We will create the model of DescisionTreeClassifier 
model = DecisionTreeClassifier(
    max_depth=3 #Maximum classfication is allowed upto 3 .
    ,
    random_state=42
)
#Here we are training the model.
model.fit(X_train,y_train)

joblib.dump(model,"student_pass_fail_model")
print("Model saved successfully")

# #predicting a new Student.
# newStudent = pd.DataFrame({
#     "hours_studied":[7],
#     "attendance":[80]
# })

# result = model.predict(newStudent)
# print(result)

# Displaying prediction.py.
```

loan approval

```python
# create model
from sklearn.tree import DecisionTreeClassifier
import pandas as pd
from sklearn.model_selection import train_test_split
import joblib
df = pd.read_csv("./documents/loan_approval.csv")

#Split into Input and Output.
X= df[["income_thousands","credit_score","existing_debt_thousands"]] #Input
print(X)

y= df['approved']
print(y)

X_train,X_test,y_train,y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42 
)
print("X_train")
print(len(X_train))
print(X_train)

#We will create the model of DescisionTreeClassifier 
model = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)
#Here we are training the model.
model.fit(X_train,y_train)

joblib.dump(model,"loan_approver")
print("Model saved successfully")
```
