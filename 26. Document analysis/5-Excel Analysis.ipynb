```python
import pandas as pd

df = pd.read_excel("./retail_sales.xlsx")#Reading from the excel file
excel_data = df.to_string(index=False)#Converting excel data --> string for sending to model
print(excel_data)

from dotenv import load_dotenv
import os
load_dotenv()
```

#### using OpenAI SDK

```python
from openai import OpenAI

client = OpenAI(
    api_key=os.getenv("openai_key")
)

while True:
    user_input = input("Ask something from excel sheet:")
    if user_input.strip().lower()=='exit':
        print("Agent : Bye Bye")
        break
    #Connecting to LLM.
    responses = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[
            {
                "role":"system",
                "content":'''
                 -You are an excel reader ai
                 -You can analysis the excel content
                 -You answer only from excel file or resources
                 -Apart from that Please Reply "I can only help with Excel Data". 
                 
                '''
            },
            {
                "role":"user",
                "content":f'''
                -Excel Data:{excel_data}
                -Question:{user_input}
                -Use Excel Data to Answer it properly .
                 
                '''
            }
        ]
    )

    msg = responses.choices[0].message.content
    print("Agent :",msg)
```

```python
from google import genai

client = genai.Client(api_key=os.getenv("gemini_key"))
print("Connected to Gemini")

while True:
    user_input = input("Ask something from excel sheet:")
    if user_input.strip().lower() == 'exit':
        print("Agent : Bye Bye")
        break

    prompt = f"-Excel Data:{excel_data}\n-Question:{user_input}\n-Use Excel Data to Answer it properly ."

    # Pass the config directly as a Python dictionary
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt,
        config={
            "system_instruction": '''
             -You are an excel reader ai
             -You can analysis the excel content
             -You answer only from excel file or resources
             -Apart from that Please Reply "I can only help with Excel Data".
            '''
        }
    )

    print("Agent :", response.text)
```
