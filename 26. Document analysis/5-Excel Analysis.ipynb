### using OpenAI SDK

```python
import pandas as pd
from openai import OpenAI
from dotenv import load_dotenv
import os

#Reading from the excel file
df = pd.read_excel("./retail_sales.xlsx")

#Converting excel data into string for sending to openai
excel_data = df.to_string(index=False)
# print(excel_data)
```

```python
load_dotenv()

#Connect to OpenAI
client = OpenAI(
    api_key=os.getenv("openai_key")
)

while True:
    user_input = input("Ask something from excel sheet:")
    if user_input.strip().lower()=='exit':
        print("Agent : Bye Bye")
        break
    #Connecting to LLM.
    responses = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[
            {
                "role":"system",
                "content":'''
                 -You are an excel reader ai
                 -You can analysis the excel content
                 -You answer only from excel file or resources
                 -Apart from that Please Reply "I can only help with Excel Data". 
                 
                '''
            },
            {
                "role":"user",
                "content":f'''
                -Excel Data:{excel_data}
                -Question:{user_input}
                -Use Excel Data to Answer it properly .
                 
                '''
            }
        ]
    )

    msg = responses.choices[0].message.content
    print("Agent :",msg)
```
