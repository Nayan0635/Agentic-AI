#### DOCX Analysis
###### install python-docx

```python
import docx

df = docx.Document("./urban_gardening.docx")
docx_data = "\n".join([p.text for p in df.paragraphs if p.text.strip()])

from dotenv import load_dotenv
import os

load_dotenv()
```

#### OpenAI SDK

```python
from openai import OpenAI

client = OpenAI(
    api_key = os.getenv("openai")
)

system_instruction = '''
act as docx file analyzer ai
answer only from provided docx file
apart from that reply: "I can't help with that.
'''

while True:
    user_input = input("You: ")
    if user_input.strip().lower() in ['exit', 'quit']:
        break
    
    prompt = f'''
    docx_data : {docx_data},
    question : {user_input},
    use the provided docx data only to answer it properly
    '''
    
    response = client.chat.completion.create(
        model = "gpt-4.1-mini",
        messages = [
            {
                "role" : "system",
                "content" : system_instruction
            },
            {
                "role" : "user_input",
                "content" : prompt
            }
        ]
    )
    msg = response.choices[0].message.content
    print("Agent: ",msg)
```

#### Gemini SDK

```python
from google import genai

client = genai.Client(
    api_key = os.getenv("gemini_key")
)

SYSTEM_CONSTRUCTION = '''
act as docx file analyzer ai
answer only from provided docx file
apart from that reply: "I can't help with that.
'''

while True:
    user_input = input("You: ")
    if user_input.strip().lower() in ['exit', 'quit']:
        break
    
    prompt = f'''
    docx_data : {docx_data},
    question : {user_input},
    use the provided docx data only to answer it properly
    '''
    
    response = client.models.generate_content(
        model = "gemini-2.5-flash-lite",
        contents = prompt,
        config={
            "system_instruction" : SYSTEM_CONSTRUCTION
        }
    )
    
    print("Agent: ",response.text)
```
