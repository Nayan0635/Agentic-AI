### CSV Analysis

```python
import pandas as pd

df = pd.read_excel("./space_missions.csv")
csv = df.to_string(index=False)#Converting excel data --> string for sending to model
print(csv)

from dotenv import load_dotenv
import os

load_dotenv()
```

#### OpenAI SDK

```python
from openai import OpenAI

client = OpenAI(
    api_key = os.getenv("openai_key")
)

system_instruction = '''
    You are csv analyzer agent, 
    answer only from provided csv file, 
    apart from that reply: I can only information available in csv file.
'''

while True:
    user_input = input("You: ")
    if user_input.strip().lower() == 'exit':
        break
    response = client.chat.completion.create(
        model = 'gpt-4.1-mini',
        messages = [
            {
                "role" : "system",
                "content" : system_instruction
            },
            {
                "role": "user",
                "content" : f'''
                    CSV data = {csv}
                    Question = {user_input}
                    use only the csv file data to answer it properly
                '''
            }
        ]
    )
    
    msg = response.choices[0].message.content
    print("Agent: ", msg)
```

#### Gemini SDK

```python
from google import genai

client = genai.Client(
    api_key = os.getenv("gemini_key")
)

system_instruction = '''
    You are csv analyzer agent, 
    answer only from provided csv file, 
    apart from that reply: I can only information available in csv file.
'''

while True:
    user_input = input("You: ")
    if user_input.strip().lower() == 'exit':
        break
    
    prompt = f'''
        CSV Data : {csv}
        Question : {user_input}
    '''
    
    response = client.models.generate_content(
        model = "gemini-3.1-flash-lite",
        contents= prompt,
        config= = {
            "system_instruction" : system_instruction
        }
    )
    print("Agent: ",response.text)
```
