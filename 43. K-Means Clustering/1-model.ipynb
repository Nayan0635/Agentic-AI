```python
from sklearn.cluster import KMeans

#selection of Input
X=[[3],[4],[10]] #Fruit size
#Create the model .
model = KMeans(n_clusters=2,random_state=42,n_init=10)
#Here we are training the Model.
model.fit(X)

#Now displaying all internally designed groups or clusters
print(model.labels_)

#Now predicting new fruit size
newFruit = [[float(input("Enter Fruit size :"))]]
result =model.predict(newFruit)
print("Belongs to Group :",result[0])
```

```python
from sklearn.cluster import KMeans

X=[[10],[12],[34],[100],[110],[210],[310]] #Fruit size
model = KMeans(n_clusters=3,random_state=42,n_init=10)
model.fit(X)

print(model.labels_)

newFruit = [[float(input("Enter Fruit size :"))]]
result =model.predict(newFruit)
print(f"{newFruit[0][0]} belongs to Group : {result[0]}")
# 182.4 -> 2 | 182.5 ->1
```
